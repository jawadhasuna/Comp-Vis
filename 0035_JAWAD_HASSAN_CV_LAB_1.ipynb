```python
# Experiment 12 — Sampling
gray = np.array(gray_image)
sample_2 = gray[::2, ::2]
sample_4 = gray[::4, ::4]
sample_8 = gray[::8, ::8]
plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1)
plt.imshow(sample_2, cmap='gray')
plt.title("Sampling: 2")
plt.axis('off')
plt.subplot(1, 3, 2)
plt.imshow(sample_4, cmap='gray')
plt.title("Sampling: 4")
plt.axis('off')
plt.subplot(1, 3, 3)
plt.imshow(sample_8, cmap='gray')
plt.title("Sampling: 8")
plt.axis('off')
plt.show()
print(gray.shape, sample_2.shape, sample_4.shape, sample_8.shape)
```

```python
# Experiment 12 — Sampling
gray = np.array(gray_image)

# Zoom into a 200x200 area from the center so the effect is visible
h, w = gray.shape
crop = gray[h//2 - 100 : h//2 + 100, w//2 - 100 : w//2 + 100]

sample_2 = crop[::2, ::2]
sample_4 = crop[::4, ::4]
sample_8 = crop[::8, ::8]

print(crop.shape, sample_2.shape, sample_4.shape, sample_8.shape)

plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(sample_2, cmap='gray', interpolation='nearest', vmin=0, vmax=255)
plt.title("Sampling: 2")
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(sample_4, cmap='gray', interpolation='nearest', vmin=0, vmax=255)
plt.title("Sampling: 4")
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(sample_8, cmap='gray', interpolation='nearest', vmin=0, vmax=255)
plt.title("Sampling: 8")
plt.axis('off')

plt.show()
```

Sampling 2 (100 × 100): the image stayed fairly clear, and most details were visible.
Sampling 4 (50 × 50): the image became rougher, and edges started to look jagged, like steps.
Sampling 8 (25 × 25): the image became very blocky, and fine details disappeared completely.

As the sampling interval increased, the number of pixels decreased and the spatial resolution dropped. Small details and sharp edges were lost first, because the samples were too far apart to capture them. The blocky, jagged appearance at low resolution is called checkerboarding or aliasing.

```python
#Task 4: Image Manipulation
#Crop the central region, resize to 256 × 256, convert to grayscale, and save the result.
img = np.array(Image.open("smile.jpg").convert('RGB'))
h, w = img.shape[:2]

# 1. Crop the central region (middle half of the image)
cropped = img[h//4 : 3*h//4, w//4 : 3*w//4]

# 2. Resize to 256 x 256
resized = cv2.resize(cropped, (256, 256), interpolation=cv2.INTER_AREA)

# 3. Convert to grayscale (the image is RGB because it came from PIL)
gray = cv2.cvtColor(resized, cv2.COLOR_RGB2GRAY)

# 4. Save the result
Image.fromarray(gray).save("processed_image.png")

# Show each step
steps = [img, cropped, resized, gray]
titles = ["Original", "Cropped", "Resized 256x256", "Grayscale"]

plt.figure(figsize=(16, 4))
for i in range(4):
    plt.subplot(1, 4, i + 1)
    if i == 3:
        plt.imshow(steps[i], cmap='gray', vmin=0, vmax=255)
    else:
        plt.imshow(steps[i])
    plt.title(titles[i])
    plt.axis('off')
plt.show()

print("Original:", img.shape)
print("Cropped:", cropped.shape)
print("Resized:", resized.shape)
print("Grayscale:", gray.shape)
```

```python
#Task 5:Sampling
#Generate image versions using sampling factors 1, 2, 4, and 8. Display and explain the differences
gray = np.array(Image.open("smile.jpg").convert('L'))
gray = cv2.resize(gray, (256, 256), interpolation=cv2.INTER_AREA)  # fixed size so the effect is clear

factors = [1, 2, 4, 8]

plt.figure(figsize=(16, 4))
for i, f in enumerate(factors, 1):
    sampled = gray[::f, ::f]
    plt.subplot(1, 4, i)
    plt.imshow(sampled, cmap='gray', interpolation='nearest', vmin=0, vmax=255)
    plt.title(f"Factor {f} ({sampled.shape[1]}x{sampled.shape[0]})")
    plt.axis('off')
    print(f"Factor {f}: {sampled.shape}, pixels = {sampled.size}")
plt.show()
```

#Factor 1: the original, with smooth face edges, curved eyes, and a clean smile.
#Factor 2: still looks good, and the difference is hard to notice.
#Factor 4: edges of the face, eyes, and smile start looking like steps.
#Factor 8: clearly blocky. The circle outline and thin curves look jagged, and small details like the shine on the face blur into blocks.



```python
#Task 6:Quantization
#Quantize a grayscale image to 256, 64, 16, 8, 4, and 2 intensity levels. Display and discuss the results.
gray = np.array(Image.open("smile.jpg").convert('L'))

def quantize_image(image, levels):
    step = 256 // levels
    return (image // step) * step

levels_list = [256, 64, 16, 8, 4, 2]

plt.figure(figsize=(15, 9))
for i, levels in enumerate(levels_list, 1):
    q = quantize_image(gray, levels)
    plt.subplot(2, 3, i)
    plt.imshow(q, cmap='gray', vmin=0, vmax=255)
    plt.title(f"{levels} levels")
    plt.axis('off')
    print(f"{levels} levels: step = {256 // levels}, distinct values in image = {len(np.unique(q))}")
plt.show()
```

56 levels, step 1: every shade stays as it is. The image uses all 256 shades, so nothing changed.
64 levels, step 4: every 4 neighboring shades become 1. For example, 0, 1, 2, 3 all become 0. That leaves 64 shades.
16 levels, step 16: every 16 shades become 1, leaving 16 shades.
8 levels, step 32: every 32 shades become 1, leaving 8 shades.
4 levels, step 64: every 64 shades become 1, leaving 4 shades (0, 64, 128, 192).
2 levels, step 128: half of all shades become 0 and the other half become 128, so only 2 shades are left: black and gray.
